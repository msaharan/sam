# 01 Data Ingestion
Fetch, inspect, and validate yfinance data for a SAM universe.

```python
from sam.config import load_universe
from sam.data import fetch_prices, validate_prices

universe = load_universe('../configs/universes/us_largecap.toml')
prices = fetch_prices(universe.normalized_symbols, start='2018-01-01', market=universe.market)
validate_prices(prices).head()
```
